```python
import heapq

def dijkstra(graph,start):
    pq = [] #우선 순위 큐
    cost = {node: float('inf') for node in graph}
    cost[start] = 0
    path = {node: start for node in graph}

    heapq.heappush(pq,(0,start))
    #path[start] = start

    while pq:
        min_val,current_node = heapq.heappop(pq)

        if min_val > cost[current_node]:
            continue

        for neighbor, weight in graph[current_node]:
            new_cost = min_val + weight
            
            if cost[neighbor] > new_cost:
                cost[neighbor] = new_cost
                heapq.heappush(pq,(new_cost,neighbor))
                path[neighbor] = path[current_node]+neighbor
    
    return path
```

```python
class Node:

    def __init__(self, value):
        self.value = value
        self.next = None


class Stack:
    def __init__(self):
        self.top = None
        self.size = 0

    def push(self,value):
        node = Node(value)
        self.size += 1
        print(self.size)

        if self.top is None:
            self.top = node
        
        else :
            node.next = self.top
            self.top = node
    
    def pop(self):
        if self.top is None:
            print('Stack is already empty')
        else :
            print(f'pop : {self.top.value}')
            self.top = self.top.next

    def peek(self):
        if self.top is None:
            print("Stack is empty")
        else :
            print(f'top : {self.top.value}')
    
    def stackSize(self):
        print(self.size)

    def isEmpty(self):
        if self.top is None:
            return True
        else:
            return False
    
    def clear(self):
        while(self.top is not None):
            self.top = self.top.next
        print('cleared')

```

```python
class Node:
    def __init__(self,value):
        self.value = value
        self.next = None

class Queue:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0
    
    def enqueue(self,value):
        node = Node(value)
        self.size += 1

        if self.head is None:
            self.head = node
            self.tail = node
        
        else:
            self.tail.next = node
            self.tail = node


    def dequeue(self):
        if self.head is None:
            print('Queue is None')
        else:    
            print(f'head : {self.head.value}')
            self.size -= 1
            if self.head == self.tail:
                self.head = None
                self.tail = None
            else:
                self.head = self.head.next

    def peek(self):
        if self.head is None:
            print('Queue is None')
        else:    
            print(f'head : {self.head.value}')        

    def qsize(self):
        print(f'Queue size : {self.size}')

    def clear(self):
        while self.head is not None:
            self.head = self.head.next
        self.tail = None
        print('Queue is cleared')
        self.size = 0

    def isEmpty(self):
        if self.head is None : 
            return True
        else:
            return False
```

```python
def merge(left,right):
    idxL = 0
    idxR = 0
    res = []
    
    while idxL < len(left) and idxR < len(right):
        if left[idxL] <= right[idxR]:
            res.append(left[idxL])
            idxL+=1
        else:
            res.append(right[idxR])
            idxR+=1
    
    for idx in range(idxL,len(left)):
        res.append(left[idx])
    
    for idx in range(idxR,len(right)):
        res.append(right[idx])
    
    return res



def mergeSort(list):
    mid = len(list)//2

    if len(list) <= 1:
        return list

    
    left = mergeSort(list[:mid])
    right = mergeSort(list[mid:])

    return merge(left,right)
```

```python
def quickSort(list):
    pivot = list[0]
    
    idxL = 0
    low = list[idxL]

    idxH=len(list)-1
    high = list[idxH]

    while idxL < idxH:
        while low < pivot:
            idxL+=1
            low = list[idxL]
        
        while pivot < high:
            idxH-=1
            high = list[idxH]
        
        list[idxL], list[idxH] = list[idxH], list[idxL]
    
    list[0],list[idxL] = list[idxL], list[0]
```

```python
def partition(list,low,high):
    pivot = high
    high -= 1

    while low <= high:

        while list[low] <= list[pivot] and low <= high:
            low+=1
        
        while list[pivot] < list[high] and low <= high:
            high-=1

        if low < high:
            list[low], list[high] = list[high], list[low]

    list[pivot], list[low] = list[low], list[pivot]

    return low

def quickSort(list,low,high):
        
        if low < high:
        
            pivot = partition(list,low,high)

            quickSort(list,low,pivot-1)
            quickSort(list,pivot+1,high)

            return list
        
        else:
             return list
```

```python
def meeting(meets):
    
    meets = meets.sort(key=lambda x:x[1])

    end_time = 0
    cnt = 0

    for start, end in meets:
        if start >= end_time:
            cnt += 1
            end_time = end
    
    print(f'최대 회의 개수: {cnt}')
```

```python
def knapsack(capacity,things):
    val = []

    for weight, price in things:
        val.append([weight,price/weight])
    
    val.sort(key=lambda x:x[1],reverse=True)

    cur = 0
    cost = 0

    print(val)

    for weight, price in val:
        if capacity-cur > weight:
            cur += weight
            cost += weight*price
        
        else:
            cost += (capacity-cur)*price
            cur = capacity


        if cur == capacity:
            break

    return cost

```
