# **1. 프롬프트 엔지니어링**
- 인공지능 모델, 특히 자연어 처리 모델과의 상호작용을 최적화하는 기술
- 모델이 의도한 답변을 도출하도록 텍스트를 구성하는 방식으로, 적절한 질문 형식이나 설명을 제공하여 더 나은 성과를 얻는 것을 목표로 함

### 1. Role
AI모델과의 대화에서 특정 문맥을 설정하거나, 모델의 응답 스타일과 관점을 조정하기 위해 사용하는 개념
- User: AI와 상호작용하는 주제로, 일반적인 질문이나 요청을 전달
- Assistant: AI 모델이 자체적으로 응답할 때의 역할로, 사용자의 질문이나 명령에 따라 답변을 제공 (예시를 만들어주거나, 미리 짜놓은 스키마를 전달할 때, 명령을 할 때 사용)
- System: AI 모델의 행동과 스타일을 지시하는 역할로, AI가 어떻게 응답해야 할지 지침을 설정

### 가격
1. GPT-X.X Sol
- 가장 성능이 좋은 모델
- 긴 코드 이해 및 생성
- 복잡한 버그 수정
- 가장 비쌈(Terra/Luna 보다 느림)
- RAG, LanGraph, 복잡한 FastAPI 프로젝트, 논문 분석, 의료AI 등

2. GPT-X.X Terra
- Sol에 가까운 성능(비용은 절반 수준)
- 성능과 비용의 균형 모델
- Python 개발, FastAPI, 코드 리뷰, 일반적인 GPT API 서비스 등

3. GPT-X.X Luna
- 매우 빠름(토큰 비용이 가장 저렴)
- 어려운 추론에서 정확도가 떨어짐
- 챗봇, 분류, 요약, 번역, 데이터 추출 등

4. input
- GPT에게 보내는 토큰
- 매 요청시

5. Cached input
- 이전에 보낸 입력을 재사용
- 캐시가 존재할 때

6. Cache writes
- 새로운 입력을 캐시에 저장
- 처음 입력했을 때

7. Output
- GPT가 생성한 답변
- 답변을 받을 때

````python
def get_response_text(response):
    for item in response.output:
        if item.type == "message":
            return item.content[0].text
    return None
````

### response.status
- completed: 응답 생성이 정상적으로 완료됨
- incomplete: 응답을 끝까지 만들지 못하고 중간에 종료됨
- failed: 오류로 인해 응답 생성 실패
- queued: 처리 대기열에 들어가 기다리는 중
- in_progress: 현재 응답을 생성하는 중
- canceled: 요청이 취소됨

````python
def handle_checkbox(selected):
    if selected:
        return "동의했습니다"
    return "동의하지 않았습니다"


with gr.Blocks() as demo:
    checkbox = gr.Checkbox(label="개인정보 사용에 동의하겠습니까?")
    output_checkbox = gr.Textbox(label="출력")
    checkbox.change(handle_checkbox,inputs=checkbox,outputs=output_checkbox)

demo.launch()
````

# **3.번역 앱 예제**

````python
# Few-shot 번역 예제 데이터
example = {
    "한국어":[
        "오늘 날씨 어때?",
        "최근 멀티모달과 Agent 기술이 인기를 끌고 있다."
    ],
    "영어":[
        "How is the weather today?",
        "Recently, multimodal and agent technologies have become popular."
    ],
    "일본어":[
        "今日の天気はどう？",
        "最近、マルチモーダルとAgent技術が人気を集めている"
    ]
}

def build_fewshot(src_lang,trg_lang):
    # 입력 언어/목표 언어의 예제
    src_examples = example[src_lang]
    trg_examples = example[trg_lang]
    fewshot_message = []

    for src_text, trg_text in zip(src_examples,trg_examples):
        fewshot_message.append({
            "role":"user", #사용자 입력
            "content":src_text
        })

        fewshot_message.append({
            "role":"assistant", #ai답변. 맥락을 기억하게 하기 위해서.
            "content":trg_text
        })

    return fewshot_message

def translate_text_chatgpt(text,src_lang,trg_lang):
    #ai에게 규칙 부여
    system_instruction = (
        "당신은 전문 번역 애플리케이션입니다.\n"
        f"입력된 {src_lang} 문장을 자연스러운 {trg_lang} 문장으로 번역하세요.\n"
        "설명, 부연 설명, 따옴표, 제목을 추가하지 마세요.\n"
        "번역된 텍스트만 출력하세요."
    )

    fewshot_messages = build_fewshot(
        src_lang=src_lang, trg_lang=trg_lang
    )

    messages = [
        {"role":"developer", "content": system_instruction},
        *fewshot_messages, #key
        {"role":"user","content":text}
    ]

    try:
        response = client.responses.create(
            model="gpt-5-nano",
            input=messages,
            reasoning={"effort":"low"}, #추론을 낮게 설정
            max_output_tokens=500
        )

        translate_text = response.output_text.strip()
        if not translate_text:
            return "번역 결과가 생성되지 않았습니다. 다시 시도해 주세요."
        return translate_text

    except Exception as e:
        print(f'OpenAI API 오류:{e}')
        return f'번역 중 오류가 발생했습니다:{e}'

def gradio_translate(text,src_lang,trg_lang):
    if text is None or not text.strip():
        return "번역할 내용을 입력하세요."
    if src_lang == trg_lang:
        return "원본 언어와 번역 언어가 같습니다. 다른 언어를 선택하세요."

    return translate_text_chatgpt(text=text.strip(),src_lang=src_lang,trg_lang=trg_lang)

with gr.Blocks() as demo:
    gr.Markdown("# 초간단 번역 앱")
    text_input = gr.Textbox(label="번역할 내용을 입력하세요",placeholder="여기에 내용을 입력하세요",lines=3)
    src_lang_dropdown = gr.Dropdown(
        choices=["한국어","영어","일본어"], value="영어",label="원본 언어를 선택하세요"
    )
    trg_lang_dropdown = gr.Dropdown(
        choices=["한국어","영어","일본어"], value="한국어",label="번역할 언어를 선택하세요"
    )
    translate_button = gr.Button("번역하기")
    output_text = gr.Textbox(
        label="번역 결과",placeholder="번역된 결과가 여기에 표시됩니다",lines=5, interactive=False
    )
    translate_button.click(
        fn=gradio_translate, inputs=[text_input,src_lang_dropdown,trg_lang_dropdown],outputs=output_text
    )
demo.launch()
````
